```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('placement.csv')
```

```python
df.head()
```

**PREPROCESSING**

```python
#removing the feature that has no impact on output
```

```python
df=df.iloc[:,1:]
```

```python
df.head()
```

**EDA**

```python
#visualising the data
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['cgpa'],df['iq'],c=df['placement'])
```

**EXTRACT INPUT AND OUTPUT COLUMNS**

```python
X=df.iloc[:,0:2]
Y=df.iloc[:,-1]
```

**TRAIN AND TEST SPLIT**

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.1)
```

**SCALING OF VALUES**

<p>Scaling is the process of bringing different feature values into a similar range so that no feature dominates the model because of its large numerical value.</p>

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_test=scaler.transform(X_test)
```

**MODEL TRAINING**

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf=LogisticRegression()
clf.fit(X_train,Y_train)
```

**MODEL EVALUATION**

```python
clf.predict(X_test)
```

```python
Y_pred=clf.predict(X_test)
```

```python
Y_test
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test,Y_pred)
```
