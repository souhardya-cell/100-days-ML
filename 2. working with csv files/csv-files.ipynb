<h1>2. Opening a csv file from an URL</h1>

<h1>3. sep & names Parameter</h1>

<p>The sep parameter in pandas pd.read_csv is used to specify the separator used between values in a file. The default sep is ',' so for comma separated files sep parameter is optional.</p>

<p>The names parameter in pandas pd.read_csv is used to specify the column names for a dataframe if it takes the first row of values as the column name by default. A list is passed with the column names written</p>

```python
pd.read_csv('movie_titles_metadata.tsv',sep='\t',names=['sno','name','release_year','imdb','votes','genre'])
```

<p>Here this movie_titles_metadata.tsv is separated by \t thats why we used that as sep.
    
The names parameter is also used to give names to the columns</p>

<h1>4. index_col parameter</h1>

```python
pd.read_csv('aug_train.csv')
```

<h1>6. usecols parameter</h1>

<h1>7. skiprows & nrows parameter</h1>

<p>skiprows parameter ignores specific rows or a number of rows while reading the dataframe.

nrows parameter only gives the specified number of rows from the dataframe</p>

```python
pd.read_csv('aug_train.csv',skiprows=[1,3],nrows=100)
```

<h1>8. encoding parameter</h1>

<p>The encoding parameter is used in pandas to specify the character encoding format of a file while reading data. Different files store text using different encoding formats. If the wrong encoding is used, pandas may show errors like:

UnicodeDecodeError

or strange unreadable characters.</p>

```python
pd.read_csv('zomato.csv')
```

```python
pd.read_csv('zomato.csv',encoding='latin-1')
```

<h1>10. Handling Dates</h1>

<p>In pandas dates are stored as objects(strings) so by default the date operations cannot be done. That is when **parse_dates** parameter is used by passing the list of column names to change its datatype to "datetime64"</p>

```python
pd.read_csv('IPL Matches 2008-2020.csv').info() #here date is object
```

```python
pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date']).info() #here date is datetime64
```

<h1>11. converters parameter</h1>

<p>The converters parameter in pandas.read_csv() is used to apply custom functions to specific columns while reading a dataset. 

Here in the below dataset we will convert the full team name of Bengaluru into their shorthand using a custom function </p>

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```

<h1>12. na_values parameter</h1>
<p>The na_values parameter in pandas.read_csv() is used to specify additional values that should be treated as missing values (NaN) while reading a dataset.</p>

```python
pd.read_csv('aug_train.csv',na_values=['Male']) #all values that are "Male" will be treated as Nan
```

<h1>13. Loading a huge dataset in chunks</h1>
<p>Loading a dataset in chunks means reading small parts of a large dataset at a time instead of loading the entire dataset into memory at once.</p>

```python
dfs = pd.read_csv('aug_train.csv',chunksize=5000)
```

<p>The chunksize parameter is used to read large datasets in smaller parts, where each chunk contains up to the specified number of rows.</p>

```python
for chunks in dfs:
    print(chunks.shape)
```
