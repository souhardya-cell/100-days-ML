<h1>Fetching Data From an API</h1>

```python
import pandas as pd
import requests
```

<h1>Loading the data from API</h1>

```python
response=requests.get('https://api.themoviedb.org/3/movie/popular?api_key=833e1ff3b6f680d67ab5b33c4b35e9e2&language=en-US&page=1')
```

<h1>Converting the data into dataframe from json</h1>

```python
response.json() #the json file has the data as a dictionary and all the data we need is assigned to the key "results", so we will only fetch that key.
```

```python
pd.DataFrame(response.json()['results']).head() #converted into DataFrame
```

<p>Now we can manipulate the data and use only those cols that are needed</p>

<h2>However this was for only 1 page and the whole dataset has 56848 pages, so we have to run a loop to merge the whole dataset, however i wont run the code as it will take long time to complete the loop for that many pages</h2>

```python
df=pd.DataFrame()
```

```python
for i in range(1,501):
    response=requests.get('https://api.themoviedb.org/3/movie/popular?api_key=833e1ff3b6f680d67ab5b33c4b35e9e2&language=en-US&page={}'.format(i))
    temp_df=pd.DataFrame(response.json()['results']).head()
    df=pd.concat([df,temp_df],ignore_index=True)
```

<p>This appends the head from every page to the newly created DataFrame "df".

Here I am only using 500 pages as the data is too big</p>

<h1>Converting the dataframe into a CSV file</h1>

```python
df.to_csv('movies.csv')
```
