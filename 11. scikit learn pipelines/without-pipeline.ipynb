```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv('train.csv')
```

```python
df.head()
```

<h3>Selecting features that are needed</h3>

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

```python
df.head()
```

<h3>Preprocessing Data</h3>

```python
#training dataset
x=df.drop(['Survived'],axis=1)
y=df['Survived']
x_train,x_test,y_train,y_test=train_test_split(x,
                                               y,
                                               test_size=0.2,
                                               random_state=42)
```

```python
x_train
```

```python
x_train.shape
```

```python
df.isnull().sum()
```

```python
#applying imputation for missing values
si_age=SimpleImputer()
si_embarked=SimpleImputer(strategy='most_frequent')

x_train_age=si_age.fit_transform(x_train[['Age']])
x_train_embarked=si_embarked.fit_transform(x_train[['Embarked']])

x_test_age=si_age.transform(x_test[['Age']])
x_test_embarked=si_embarked.transform(x_test[['Embarked']])
```

```python
#ohe sex and embarked
ohe_sex=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
ohe_embarked=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
#as we are using Decison Trees and thats why we dont need to worry about multicollinearity.

x_train_sex=ohe_sex.fit_transform(x_train[['Sex']])
x_train_embarked=ohe_embarked.fit_transform(x_train_embarked)

x_test_sex=ohe_sex.transform(x_test[['Sex']])
x_test_embarked=ohe_embarked.transform(x_test_embarked)
```

```python
#extracting the remaining data
x_train_rem=x_train.drop(['Sex','Age','Embarked'],axis=1)
x_test_rem=x_test.drop(['Sex','Age','Embarked'],axis=1)
```

```python
#concatenating all of it
x_train_transformed=np.concatenate((x_train_rem,x_train_sex,x_train_age,x_train_embarked),axis=1)
x_test_transformed=np.concatenate((x_test_rem,x_test_sex,x_test_age,x_test_embarked),axis=1)
```

```python
x_test_transformed.shape
```

<h1>APPLYING DECISION TREE ALGO</h1>

```python
clf=DecisionTreeClassifier()
clf.fit(x_train_transformed,y_train)
y_pred=clf.predict(x_test_transformed)
```

```python
y_pred
```

<h1>Finding Accuracy</h1>

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

<h1>Dumping using pickle</h1>

pickle.dump() serializes and saves trained ML objects into files so they can be loaded and reused later without retraining.</p>

```python
import pickle
```

```python
pickle.dump(ohe_sex,open('models/ohe_sex.pkl','wb'))
pickle.dump(ohe_embarked,open('models/ohe_embarked.pkl','wb'))
pickle.dump(clf,open('models/clf.pkl','wb'))
```
