```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

<h1>Working on number column</h1>

```python
# extract numerical part
df['number_numerical'] = pd.to_numeric(df["number"],errors='coerce',downcast='integer')
```

```python
#errors='coerce' invalid values are replaced with NaN instead of throwing an error.
```

```python
# extract categorical part
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['number'],np.nan)
```

```python
# understanding np.where():
#np.where(condition, value_if_true, value_if_false)
#if condition is True → take first value
#otherwise → take second value
```

```python
df.head()
```

<h1>Working on Cabin column</h1>

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)') # captures numerical part
df['cabin_cat'] = df['Cabin'].str[0] # captures the first letter
```

```python
#(\d+) means extract continuous numbers
#.str[0] extracts the first character of each string
```

```python
df.head()
```

<h1>Working on Ticket column</h1>

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])
```

```python
#split() divides a string into parts using spaces,
#and then its taking the last part of the ticket as the numeric part.

#the second line converts extracted ticket numbers into numerical datatype.

#the third line takes first word from each ticket into catergory.

#however, some tickets are purely numeric(index=3,4), first part is also numeric.

#so np.where(condition, true_value, false_value)
#if condition is True → take first value
#otherwise → take second value
```
