<h1>Working with Dates</h1>

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

<h3>1. Extract year</h3>

```python
date['date_year'] = date['date'].dt.year
date.sample(5)
```

<h3>2. Extract Month</h3>

```python
date['date_month_no'] = date['date'].dt.month
date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.head()
```

<h3>3. Extract Days</h3>

```python
date['date_day'] = date['date'].dt.day
date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek
date.head()
```

```python
# day of week - name
date['date_dow_name'] = date['date'].dt.day_name()
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)
date.drop(columns=['product_id','city_id','orders']).head()
```

<h3>4. Extract Quarter</h3>

```python
date['quarter'] = date['date'].dt.quarter
date.drop(columns=['product_id','city_id','orders']).head()
```

<h3>5. Extract Semester</h3>

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)
date.drop(columns=['product_id','city_id','orders']).head()
```

<h3>6. Extract Time elapsed between dates</h3>

```python
import datetime
today=datetime.datetime.today()
```

```python
today
```

```python
#time passed
today - date['date']
```

```python
#days passed
(today - date['date']).dt.days
```

```python
# Months passed

np.round((today -date['date']).dt.days/30,0)
```

<h1>Working with Time</h1>

```python
# Converting to datetime datatype
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

<h3>Extract Time part</h3>

```python
time['time'] = time['date'].dt.time
time.head()
```

<h3>Time Difference</h3>

```python
today-time['date']
```

```python
#in seconds
np.round((today-time['date']).dt.total_seconds())
```

```python
#in minutes
np.round((today-time['date']).dt.total_seconds()/60)
```

```python
#in hours
np.round((today-time['date']).dt.total_seconds()/3600)
```
