```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes

data=load_diabetes()
```

```python
X=data.data
y=data.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
from sklearn.linear_model import LinearRegression
L=LinearRegression()
```

```python
L.fit(X_train,y_train)
```

```python
print(L.coef_)
print(L.intercept_)
```

```python
y_pred=L.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error

print("R2 score",r2_score(y_test,y_pred))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

<h1>Ridge Regression</h1>

```python
from sklearn.linear_model import Ridge
R=Ridge(alpha=0.01)
```

```python
R.fit(X_train,y_train)
```

```python
print(R.coef_)
print(R.intercept_)
```

```python
y_pred1=R.predict(X_test)
```

```python
print("R2 score",r2_score(y_test,y_pred1))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred1)))
```

<h1>On a Polynomial Dataset</h1>

```python
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```

<p>The graph shows how different values of alpha affect a degree-16 Polynomial Regression model. The blue plus signs are the actual data points. When alpha = 0, there is no regularization, so the model behaves like ordinary Linear Regression after polynomial features are created. Because degree 16 makes the model very flexible, the red curve follows the training data too closely and can overfit. When alpha = 20, some regularization is applied, so the curve becomes smoother and more controlled, giving a better balance. When alpha = 200, the regularization is very strong, shrinking the coefficients too much, so the blue curve becomes overly smooth and fails to capture the pattern properly, resulting in underfitting. Overall, the graph shows that choosing a suitable alpha is important: too little regularization can cause overfitting, while too much can cause underfitting.
</p>
