```python
from sklearn.datasets import load_diabetes

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
data = load_diabetes()
```

```python
df = pd.DataFrame(data.data,columns=data.feature_names)
df['TARGET'] = data.target
```

```python
df.head()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(data.data,data.target,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
```

<h1>How are coefficients affected?</h1>

```python
coefs = []
r2_scores = []

for i in [0,10,100,1000]:
    reg = Ridge(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())
    y_pred = reg.predict(X_test)
    r2_scores.append(r2_score(y_test,y_pred))
```

```python
plt.figure(figsize=(14,9))
plt.subplot(221)
plt.bar(data.feature_names,coefs[0])
plt.title('Alpha = 0 ,r2_score = {}'.format(round(r2_scores[0],2)))

plt.subplot(222)
plt.bar(data.feature_names,coefs[1])
plt.title('Alpha = 10 ,r2_score = {}'.format(round(r2_scores[1],2)))

plt.subplot(223)
plt.bar(data.feature_names,coefs[2])
plt.title('Alpha = 100 ,r2_score = {}'.format(round(r2_scores[2],2)))

plt.subplot(224)
plt.bar(data.feature_names,coefs[3])
plt.title('Alpha = 1000 ,r2_score = {}'.format(round(r2_scores[3],2)))

plt.show()
```

### How Ridge Changes the Coefficients

The four coefficient plots show how increasing $\alpha$ changes the learned coefficients of Ridge Regression. When $\alpha = 0$, the model behaves like ordinary Linear Regression and some coefficients have relatively large positive or negative values. As $\alpha$ increases to $10$, $100$, and $1000$, the coefficients are progressively pushed closer to zero. The $R^2$ score also changes as the amount of regularization increases, showing that stronger regularization changes the model's fit to the data. The main idea is that Ridge controls the size of the coefficients rather than forcing them to become zero.

```python
alphas = [0,0.0001,0.0005,0.001,0.005,0.1,0.5,1,5,10]

coefs = []

for i in alphas:
    reg = Ridge(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())
```

```python
input_array = np.array(coefs).T
```

```python
plt.figure(figsize=(15,8))
plt.plot(alphas,np.zeros(len(alphas)),color='black',linewidth=5)
for i in range(input_array.shape[0]):
    plt.plot(alphas,input_array[i],label=data.feature_names[i])
plt.legend()
```

### Higher Coefficients Are Affected More

The coefficient-path graph shows how each feature's coefficient changes as the Ridge parameter $\alpha$ increases. Features that initially have large coefficient magnitudes experience a stronger reduction as regularization becomes stronger, while coefficients that are already small change comparatively less. This happens because Ridge uses an L2 penalty, $\lambda\sum \beta_j^2$, so larger coefficients contribute much more to the regularization cost. The graph therefore demonstrates that Ridge particularly controls large coefficient values and gradually shrinks them toward zero.

<h1>3. Impact on Bias and Variance</h1>

```python
m = 100
X = 5 * np.random.rand(m, 1) - 2
y = 0.7 * X ** 2 - 2 * X + 3 + np.random.randn(m, 1)

plt.scatter(X, y)
plt.show()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X.reshape(100,1),y.reshape(100),test_size=0.2,random_state=2)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=15)

X_train = poly.fit_transform(X_train)
X_test = poly.transform(X_test)
```

```python
from mlxtend.evaluate import bias_variance_decomp

alphas = np.linspace(0,30,100)

loss = []
bias = []
variance = []

for i in alphas:
    reg = Ridge(alpha=i)
    avg_expected_loss, avg_bias, avg_var = bias_variance_decomp(
        reg, X_train, y_train, X_test, y_test, 
        loss='mse',
        random_seed=123)
    loss.append(avg_expected_loss)
    bias.append(avg_bias)
    variance.append(avg_var)
```

```python
plt.plot(alphas,loss,label='loss')
plt.plot(alphas,bias,label='Bias')
plt.plot(alphas,variance,label='Variance')
plt.ylim(0,5)
plt.xlabel('Alpha')
plt.legend()
```

### Impact on Bias and Variance

This graph demonstrates the effect of increasing $\alpha$ on loss, bias, and variance using a high-degree polynomial model. With little or no regularization, the model can become overly sensitive to the training data, leading to higher variance. As $\alpha$ increases, Ridge constrains the coefficients and reduces this variance, but at the same time the added constraint introduces more bias. The loss therefore reflects the trade-off between these two effects. The main idea is that Ridge regularization can control an overly flexible model by exchanging some bias for a reduction in variance.

<h1>4. Effect of Regularization on Loss Function</h1>

```python
from sklearn.datasets import make_regression

X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)

plt.scatter(X,y)

from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X,y)
print(reg.coef_)
print(reg.intercept_)
```

```python
def cal_loss(m,alpha):
    return np.sum((y - m*X.ravel() + 2.29)**2) + alpha*m*m
```

```python
def predict(m):
    return m*X - 2.29
```

```python
m = np.linspace(-45,100,100)
plt.figure(figsize=(4,6))
for j in [0,10,20,30,40,50,100]:
    loss = []
    for i in range(m.shape[0]):
        loss_i = cal_loss(m[i],j)
        loss.append(loss_i)
    plt.plot(m,loss,label='alpha = {}'.format(j))
plt.legend()
plt.xlabel('m')
plt.ylabel('Loss')
plt.show()
```

### Effect of Regularization on the Loss Function

The final graph shows how adding the Ridge penalty changes the loss for different values of $\alpha$. The loss contains both the ordinary squared-error term and the regularization term $\alpha m^2$. When $\alpha = 0$, the loss depends only on the fitting error. As $\alpha$ increases, large values of the coefficient $m$ become more expensive, so the loss curve is modified to discourage large coefficients. This shifts the optimization toward solutions with smaller parameter values, illustrating how Ridge changes the objective that the model is trying to minimize.
